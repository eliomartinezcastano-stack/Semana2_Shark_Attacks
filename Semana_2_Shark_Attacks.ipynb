Seleccionamos columnas relevantes para nuestra hipótesis.

```python
Shark_Attacks_df.columns = (Shark_Attacks_df.columns.str.strip().str.lower().str.replace(" ", "_"))
Shark_Attacks_df = Shark_Attacks_df.rename(columns={"fatal_y/n": "fatal"})
Shark_Attacks_df.columns
```

```python
Shark_Attacks_df = Shark_Attacks_df[["country", "activity", "injury", "fatal"]]
Shark_Attacks_df.head()
```

Exploramos valores únicos

```python
for col in Shark_Attacks_df:
    print(col)
    print("Valores únicos: ", Shark_Attacks_df[col].unique())
    print("Número de valores únicos: ", Shark_Attacks_df[col].nunique())
```

```python
for col in Shark_Attacks_df:
    Shark_Attacks_df[col] = Shark_Attacks_df[col].astype(str).str.lower().str.strip()

Shark_Attacks_df
```

Nos quedamos con el primer país en ataques que han sucedido entre 2 países.

```python
Shark_Attacks_df['country'] = (Shark_Attacks_df['country'].str.split('/').str[0])
```

```python
Shark_Attacks_df['country'] = (Shark_Attacks_df['country'].str.replace(r"[\.\/\?\(\)]", "", regex=True))
```

```python
Shark_Attacks_df['country'].value_counts().head(20)
```

```python
Shark_Attacks_df['country'].nunique()
```

```python
print("Top 3 países tras la limpieza:")
top_3_paises = (Shark_Attacks_df['country'].value_counts().head(3).index.tolist())
print (top_3_paises)
```

```python
df_top3 = Shark_Attacks_df[Shark_Attacks_df['country'].isin(top_3_paises)]
df_top3
```

```python
import re
df_top3['activity_group'] = df_top3['activity'].apply(
    lambda x: "surfing/boarding" if re.search(r"surf|boarding|bodyboard|paddle", x)
    else "swimming" if re.search(r"swim|bath|float|treading", x)
    else "fishing" if re.search(r"ish|spear|net", x)
    else "diving" if re.search(r"dive|snorkel|scuba", x)
    else "boating" if re.search(r"kayak|canoe|row|boat|sail", x)
    else "other")

df_top3['activity_group'].unique()
```

```python
#Viendo cuales pueden ser los valores más repetidos, los vamos a clasificar en partes generales del cuerpo.
```

```python
df_top3['injury_group'] = df_top3['injury'].apply(
    lambda x: "head" if re.search(r"head|face|eye|scalp|mouth|neck", x)
    else "arm" if re.search(r"arm|hand|finger|shoulder|elbow|wrist", x)
    else "leg" if re.search(r"leg|thigh|foot|toe|knee|ankle|calf", x)
    else "torso" if re.search(r"torso|chest|back|abdomen|stomach", x)
    else "other")

df_top3['injury_group'].unique()
```
